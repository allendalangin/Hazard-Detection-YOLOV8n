```python
# Import the PyTorch library
import torch

def check_gpu():
    """
    This function checks if a CUDA-compatible GPU is available and usable by PyTorch.
    """
    print("🚀 Running GPU Check...")
    
    # Check if CUDA is available
    if torch.cuda.is_available():
        # Get the number of available GPUs
        gpu_count = torch.cuda.device_count()
        print(f"✅ Success! Found {gpu_count} CUDA-enabled GPU(s).")
        
        # Get the name of the current GPU (device 0)
        gpu_name = torch.cuda.get_device_name(0)
        print(f"   - Device 0: {gpu_name}")
        
        # Get the CUDA version PyTorch was compiled with
        pytorch_cuda_version = torch.version.cuda
        print(f"   - PyTorch is using CUDA version: {pytorch_cuda_version}")

    else:
        print("❌ Failed. PyTorch cannot find a CUDA-enabled GPU.")
        print("   - Please ensure your NVIDIA drivers and CUDA Toolkit are installed correctly.")

if __name__ == "__main__":
    check_gpu()
```

```python
import os

# --- CONFIGURATION ---
# The path to your final dataset folder
dataset_base_path = './cleaned_hazard_dataset/'
# --- END OF CONFIGURATION ---

def count_images_in_split(split_name):
    """Counts the number of files in a specific split's images directory."""
    images_path = os.path.join(dataset_base_path, split_name, 'images')
    
    # Check if the directory exists to prevent errors
    if not os.path.exists(images_path):
        print(f"Directory not found: {images_path}")
        return 0
        
    # List all files in the directory and return the count
    return len(os.listdir(images_path))

# --- MAIN SCRIPT ---
print(f"🔍 Counting images in '{dataset_base_path}'...")

train_count = count_images_in_split('train')
valid_count = count_images_in_split('valid')
test_count = count_images_in_split('test')

total_count = train_count + valid_count + test_count

print("-" * 30)
print(f"Training set images:   {train_count}")
print(f"Validation set images: {valid_count}")
print(f"Testing set images:    {test_count}")
print("-" * 30)
print(f"Total images:          {total_count}")
print("-" * 30)

# Calculate and print percentages
if total_count > 0:
    train_percent = (train_count / total_count) * 100
    valid_percent = (valid_count / total_count) * 100
    test_percent = (test_count / total_count) * 100
    
    print("Dataset Distribution:")
    print(f"  - Train: {train_percent:.1f}%")
    print(f"  - Valid: {valid_percent:.1f}%")
    print(f"  - Test:  {test_percent:.1f}%")
```
